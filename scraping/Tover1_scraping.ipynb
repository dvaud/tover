```python
import requests

url = "https://www.nhs.uk/conditions/dementia/living-with-dementia/activities/"

response = requests.get(url, timeout=30)

print(response.status_code)
response.raise_for_status()
```

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(response.text, "html.parser")

title = soup.select("main h1")

print(title[0].get_text(strip=True))
```

```python
content = soup.select("main article p, main article li")

print(len(content))
```

```python
for i, item in enumerate(content[:5]):
    print(f"{i + 1}. {item.get_text(' ', strip=True)}")
    print()
```

```python
elements = soup.select(
    "main article h2, main article h3, main article p, main article li"
)

data = []
section = "Introduction"
subsection = ""

for item in elements:
    text = item.get_text(" ", strip=True)

    if item.name == "h2":
        section = text
        subsection = ""

    elif item.name == "h3":
        subsection = text

    else:
        data.append({
            "order": len(data) + 1,
            "section": section,
            "subsection": subsection,
            "tag": item.name,
            "text": text,
            "source_url": url
        })
```

```python
import pandas as pd

df = pd.DataFrame(data)

print("Number of rows:", len(df))

df.head()
```

```python
print(df["tag"].value_counts())
```

```python
print("Empty texts:", (df["text"].str.strip() == "").sum())
```

```python
print(df.loc[0, "text"])
df.tail()
```

```python
df.to_csv("nhs_activities_dementia.csv", index=False)

from pathlib import Path

print(Path("nhs_activities_dementia.csv").resolve())
```
